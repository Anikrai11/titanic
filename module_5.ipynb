# question 2


```python
import numpy as np
from scipy import stats
data=[1,2,2,3,42]
mean_value=np.mean(data)
median_value=np.median(data)
mode_reasult=stats.mode(data,keepdims=True)
print("Mean:",mean_value)
print("Median:",median_value)
print("Mode:",mode_reasult)
```

```python
import numpy as np
minute_on_site=[5,10,15,20,25,30]
money_spent=[12,22,35,48,55,66]
correlation=np.corrcoef(minute_on_site,money_spent)[1,0]
print(round(correlation,2))
```

## question 5


```python
import numpy as np
from scipy import stats
mean=0
std=1
lower=mean-1*std
upper=mean+1*std
percentage=(stats.norm.cdf(upper)-stats.norm.cdf(lower))*100
print(round(percentage,2))
```
